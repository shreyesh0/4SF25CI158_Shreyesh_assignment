```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)),
                  columns=['A', 'B', 'C'])

df['Product'] = df['A'] * df['B']

print(df)
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (4, 3)),
                  columns=['A', 'B', 'C'])

print(df)

print("Row-wise Sum:")
print(df.sum(axis=1))

print("Column-wise Sum:")
print(df.sum(axis=0))
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)),
                  columns=['A', 'B', 'C'])

df.iloc[1, 0] = np.nan
df.iloc[2, 1] = np.nan
df.iloc[3, 2] = np.nan

print("Before:")
print(df)

df = df.fillna(df.mean())

print("After:")
print(df)
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (6, 4)),
                  columns=['A', 'B', 'C', 'D'])

df.iloc[1, 0] = np.nan
df.iloc[3, 2] = np.nan

print("Before:")
print(df)

df = df.dropna()

print("After:")
print(df)
```

```python
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], 10),
    'Value': np.random.randint(10, 100, 10)
})

print(df)

result = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print(result)
```

```python
df = pd.DataFrame({
    'Product': np.random.choice(['Pen', 'Book', 'Bag'], 10),
    'Category': np.random.choice(['A', 'B', 'C'], 10),
    'Sales': np.random.randint(100, 1000, 10)
})

print(df)

result = df.groupby('Category')['Sales'].sum()

print(result)
```

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Amit', 'Ravi', 'Rahul']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [80, 90, 85]
})

result = pd.merge(df1, df2, on='ID')

print(result)
```

```python
index = pd.MultiIndex.from_tuples([
    ('A', 1), ('A', 2),
    ('B', 1), ('B', 2),
    ('C', 1), ('C', 2)
], names=['Category', 'Number'])

df = pd.DataFrame({'Value': [10, 20, 30, 40, 50, 60]},
                  index=index)

print(df)

print("Basic Indexing:")
print(df.loc['A'])

print("Specific Element:")
print(df.loc[('B', 1)])

print("Slicing:")
print(df.loc['A':'B'])
```

```python
df = pd.DataFrame({
    'Category': ['A', 'A', 'A', 'B', 'B', 'B'],
    'SubCategory': ['X', 'Y', 'X', 'X', 'Y', 'Y'],
    'Value': np.random.randint(10, 100, 6)
})

df = df.set_index(['Category', 'SubCategory'])

print(df)

result = df.groupby(level=['Category', 'SubCategory']).sum()

print(result)
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)),
                  columns=['A', 'B', 'C'])

print("Original:")
print(df)

result = df.apply(lambda x: x * 2)

print("Doubled:")
print(result)
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (6, 3)),
                  columns=['A', 'B', 'C'])

df['Sum'] = df.apply(lambda x: x['A'] + x['B'] + x['C'], axis=1)

print(df)
```
